```python
import numpy as np
from google.colab.patches import cv2_imshow

# 5x7 font using '1' and '0'
CHAR_WIDTH = 5
CHAR_HEIGHT = 7

FONT = {
    "A": [
        "00100",
        "01010",
        "10001",
        "11111",
        "10001",
        "10001",
        "10001"
    ],
    "B": [
        "11110",
        "10001",
        "10001",
        "11110",
        "10001",
        "10001",
        "11110"
    ],
    "C": [
        "01111",
        "10000",
        "10000",
        "10000",
        "10000",
        "10000",
        "01111"
    ],
    "D": [
        "11100",
        "10010",
        "10001",
        "10001",
        "10001",
        "10010",
        "11100"
    ],
    "E": [
        "11111",
        "10000",
        "10000",
        "11110",
        "10000",
        "10000",
        "11111"
    ],
    "F": [
        "11111",
        "10000",
        "10000",
        "11110",
        "10000",
        "10000",
        "10000"
    ],
    "G": [
        "01111",
        "10000",
        "10000",
        "10011",
        "10001",
        "10001",
        "01111"
    ],
    "H": [
        "10001",
        "10001",
        "10001",
        "11111",
        "10001",
        "10001",
        "10001"
    ],
    "I": [
        "01110",
        "00100",
        "00100",
        "00100",
        "00100",
        "00100",
        "01110"
    ],
    "J": [
        "00111",
        "00010",
        "00010",
        "00010",
        "10010",
        "10010",
        "01100"
    ],
    "K": [
        "10001",
        "10010",
        "10100",
        "11000",
        "10100",
        "10010",
        "10001"
    ],
    "L": [
        "10000",
        "10000",
        "10000",
        "10000",
        "10000",
        "10000",
        "11111"
    ],
    "M": [
        "10001",
        "11011",
        "10101",
        "10001",
        "10001",
        "10001",
        "10001"
    ],
    "N": [
        "10001",
        "11001",
        "10101",
        "10011",
        "10001",
        "10001",
        "10001"
    ],
    "O": [
        "01110",
        "10001",
        "10001",
        "10001",
        "10001",
        "10001",
        "01110"
    ],
    "P": [
        "11110",
        "10001",
        "10001",
        "11110",
        "10000",
        "10000",
        "10000"
    ],
    "Q": [
        "01110",
        "10001",
        "10001",
        "10001",
        "10101",
        "10010",
        "01101"
    ],
    "R": [
        "11110",
        "10001",
        "10001",
        "11110",
        "10100",
        "10010",
        "10001"
    ],
    "S": [
        "01111",
        "10000",
        "10000",
        "01110",
        "00001",
        "00001",
        "11110"
    ],
    "T": [
        "11111",
        "00100",
        "00100",
        "00100",
        "00100",
        "00100",
        "00100"
    ],
    "U": [
        "10001",
        "10001",
        "10001",
        "10001",
        "10001",
        "10001",
        "01110"
    ],
    "V": [
        "10001",
        "10001",
        "10001",
        "10001",
        "10001",
        "01010",
        "00100"
    ],
    "W": [
        "10001",
        "10001",
        "10001",
        "10001",
        "10101",
        "11011",
        "10001"
    ],
    "X": [
        "10001",
        "10001",
        "01010",
        "00100",
        "01010",
        "10001",
        "10001"
    ],
    "Y": [
        "10001",
        "10001",
        "01010",
        "00100",
        "00100",
        "00100",
        "00100"
    ],
    "Z": [
        "11111",
        "00001",
        "00010",
        "00100",
        "01000",
        "10000",
        "11111"
    ],
    " ": [
        "00000",
        "00000",
        "00000",
        "00000",
        "00000",
        "00000",
        "00000"
    ],
    ".": [
        "00000",
        "00000",
        "00000",
        "00000",
        "00000",
        "01100",
        "01100"
    ],
    ",": [
        "00000",
        "00000",
        "00000",
        "00000",
        "01100",
        "01100",
        "00100"
    ],
    "!": [
        "00100",
        "00100",
        "00100",
        "00100",
        "00100",
        "00000",
        "00100"
    ],
    "?": [
        "01110",
        "10001",
        "00001",
        "00010",
        "00100",
        "00000",
        "00100"
    ],
    '"': [
        "01010",
        "01010",
        "00000",
        "00000",
        "00000",
        "00000",
        "00000"
    ],
    "#": [
        "01010",
        "11111",
        "01010",
        "01010",
        "11111",
        "01010",
        "01010"
    ],
    "$": [
        "00100",
        "01111",
        "10100",
        "01110",
        "00101",
        "11110",
        "00100"
    ],
    "%": [
        "11000",
        "11001",
        "00010",
        "00100",
        "01000",
        "10011",
        "00011"
    ],
    "&": [
        "01100",
        "10010",
        "10100",
        "01000",
        "10101",
        "10010",
        "01101"
    ],
    "'": [
        "00100",
        "00100",
        "00000",
        "00000",
        "00000",
        "00000",
        "00000"
    ],
    "(": [
        "00010",
        "00100",
        "01000",
        "01000",
        "01000",
        "00100",
        "00010"
    ],
    ")": [
        "01000",
        "00100",
        "00010",
        "00010",
        "00010",
        "00100",
        "01000"
    ],
    "*": [
        "00000",
        "01010",
        "00100",
        "11111",
        "00100",
        "01010",
        "00000"
    ],
    "+": [
        "00000",
        "00100",
        "00100",
        "11111",
        "00100",
        "00100",
        "00000"
    ],
    ",": [
        "00000",
        "00000",
        "00000",
        "00000",
        "01100",
        "01100",
        "00100"
    ],
    "-": [
        "00000",
        "00000",
        "00000",
        "11111",
        "00000",
        "00000",
        "00000"
    ],
    "/": [
        "00001",
        "00010",
        "00100",
        "01000",
        "10000",
        "00000",
        "00000"
    ],
    ":": [
        "00000",
        "01100",
        "01100",
        "00000",
        "01100",
        "01100",
        "00000"
    ],
    ";": [
        "00000",
        "01100",
        "01100",
        "00000",
        "01100",
        "01100",
        "00100"
    ],
    "<": [
        "00010",
        "00100",
        "01000",
        "10000",
        "01000",
        "00100",
        "00010"
    ],
    "=": [
        "00000",
        "11111",
        "00000",
        "11111",
        "00000",
        "00000",
        "00000"
    ],
    ">": [
        "10000",
        "01000",
        "00100",
        "00010",
        "00100",
        "01000",
        "10000"
    ],
    "@": [
        "01110",
        "10001",
        "10011",
        "10101",
        "10111",
        "10000",
        "01110"
    ],
    "[": [
        "01110",
        "01000",
        "01000",
        "01000",
        "01000",
        "01000",
        "01110"
    ],
    "\\": [
        "10000",
        "01000",
        "00100",
        "00010",
        "00001",
        "00000",
        "00000"
    ],
    "]": [
        "01110",
        "00010",
        "00010",
        "00010",
        "00010",
        "00010",
        "01110"
    ],
    "^": [
        "00100",
        "01010",
        "10001",
        "00000",
        "00000",
        "00000",
        "00000"
    ],
    "_": [
        "00000",
        "00000",
        "00000",
        "00000",
        "00000",
        "00000",
        "11111"
    ],
    "`": [
        "00100",
        "00100",
        "00010",
        "00000",
        "00000",
        "00000",
        "00000"
    ],
    "{": [
        "00010",
        "00100",
        "00100",
        "01000",
        "00100",
        "00100",
        "00010"
    ],
    "|": [
        "00100",
        "00100",
        "00100",
        "00100",
        "00100",
        "00100",
        "00100"
    ],
    "}": [
        "01000",
        "00100",
        "00100",
        "00010",
        "00100",
        "00100",
        "01000"
    ],
    "~": [
        "01010",
        "10101",
        "00000",
        "00000",
        "00000",
        "00000",
        "00000"
    ]
}


def get_char_bitmap(char):
    """Returns 2D numpy array (0, 1) for given character"""
    bitmap = FONT.get(char, FONT[" "])
    return np.array([[int(px) for px in row] for row in bitmap], dtype=np.uint8)

def render_text_opencv(text, scale=5, spacing=1, line_spacing=2, left_padding=1, top_padding=1):
    text = text.upper().split('\n')
    text = [(' ' * left_padding + line) for line in text]  # Apply left padding

    max_len = max(len(line) for line in text)

    # Character + spacing dimensions
    char_w = (CHAR_WIDTH + spacing) * scale
    char_h = (CHAR_HEIGHT) * scale
    img_w = max_len * char_w
    img_h = (len(text) + top_padding) * (char_h + line_spacing * scale)

    # Create blank white image
    canvas = np.ones((img_h, img_w), dtype=np.uint8) * 255

    for row_idx, line in enumerate(text):
        for col_idx, char in enumerate(line):
            bitmap = get_char_bitmap(char)
            for y in range(CHAR_HEIGHT):
                for x in range(CHAR_WIDTH):
                    if bitmap[y, x] == 1:
                        top_left_y = (row_idx + top_padding) * (char_h + line_spacing * scale) + y * scale
                        top_left_x = col_idx * char_w + x * scale
                        canvas[top_left_y:top_left_y+scale, top_left_x:top_left_x+scale] = 0  # Black square

    # Show image
    cv2_imshow(canvas)

# Example usage
if __name__ == "__main__":
    print("Enter your text (type 'END' on a new line to finish):")
    lines = []
    while True:
        line = input()
        if line.strip().upper() == 'END':
            break
        lines.append(line)

    user_input = "\n".join(lines)
    render_text_opencv(user_input)
```
